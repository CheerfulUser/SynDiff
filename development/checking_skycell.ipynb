```python
import numpy as np
import matplotlib.pyplot as plt

from astropy.io import fits

%matplotlib widget
```

```python
file = '/home/phys/astronomy/zgl12/SkyCells/New_SC/Combined_Data/rings.v3.skycell.2556.087.stk.rizy.conv.fits'
```

```python
hdul = fits.open(file)
data = hdul[0].data
hdul.close()

plt.figure()
plt.imshow(data, origin='lower', cmap='gray', vmin=np.nanpercentile(data, 15), vmax=np.nanpercentile(data, 85))
plt.colorbar()
plt.title('Skycell Data Visualization')
plt.xlabel('X Pixel')
plt.ylabel('Y Pixel')
plt.tight_layout()
plt.savefig('skycell_visualization.png', dpi=300)
plt.show()
```
